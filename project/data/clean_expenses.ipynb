```python
# ==============================
# Clean Expenses Notebook
# ==============================

import pandas as pd

# 1) Load data
df = pd.read_csv("expenses.csv")

print("Original shape:", df.shape)
df.head()
```

```python
# 2) Basic column check
required_cols = ['date', 'category', 'amount']

for col in required_cols:
    if col not in df.columns:
        raise ValueError(f"Missing required column: {col}")

print("All required columns exist.")
```

```python
# 3) Clean date column
df['date'] = pd.to_datetime(df['date'], errors='coerce')
df = df.dropna(subset=['date'])
```

```python
# 4) Validate amount column
def validate_amount(x):
    try:
        x = float(x)
        return x if x > 0 else None
    except:
        return None

df['amount'] = df['amount'].apply(validate_amount)
df = df.dropna(subset=['amount'])
```

```python
# 5) Normalize category
df['category'] = df['category'].astype(str).str.strip().str.lower()
```

```python
# 6) Derived columns
df['day_of_week'] = df['date'].dt.day_name()
df['week'] = df['date'].dt.isocalendar().week
df['is_weekend'] = df['date'].dt.weekday >= 5
```

```python
# 7) Final check
print("Cleaned shape:", df.shape)
df.head()
```

```python
# 8) Save clean data
df.to_csv("clean_expenses.csv", index=False)

print("clean_expenses.csv saved successfully.")
```
