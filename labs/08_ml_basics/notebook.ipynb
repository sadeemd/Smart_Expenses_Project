Preparing Features and Labels (X, y) for Machine Learning

```python
import pandas as pd

# -----------------------------
# 1) Create a simple dataset
# -----------------------------
df = pd.DataFrame({
    "day": [1, 2, 3, 4, 5, 6, 7],
    "category": ["Food", "Transport", "Food", "Bills", "Food", "Other", "Transport"],
    "amount": [12000, 5000, 8000, 25000, 7000, 4000, 6000],
    "weekly_total": [67000, 67000, 67000, 67000, 67000, 67000, 67000]  # مثال بسيط
})

df
```

```python
# -----------------------------
# 2) Choose features
# -----------------------------
# Features: day + category
X = df[["day", "category"]]

# Convert category to one-hot
X = pd.get_dummies(X, columns=["category"])

X.head()
```

```python
# -----------------------------
# 2) Choose features
# -----------------------------
# Features: day + category
X = df[["day", "category"]]

# Convert category to one-hot
X = pd.get_dummies(X, columns=["category"])
X = X.astype(int)
X.head()
```

```python
# -----------------------------
# 3) Choose label (target)
# -----------------------------
y = df["amount"]

y.head()
```

```python
# -----------------------------
# 4) Split X, y
# -----------------------------
# X already prepared, y already prepared
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# -----------------------------
# 5) Validate X and y
# -----------------------------
print("Missing values in X:", X.isna().sum().sum())
print("Missing values in y:", y.isna().sum())

print("\nX columns:", list(X.columns))
```
