# Day 5 — Matplotlib Basics
Goal: Create a simple line plot and bar chart, add labels/title/grid, and save figures as PNG files.


```python
import matplotlib.pyplot as plt
```

## Line Plot — 7 Days Expenses
We will use simple sample data for daily expenses during one week.


```python
days = ["Sat", "Sun", "Mon", "Tue", "Wed", "Thu", "Fri"]
expenses = [10, 15, 8, 20, 12, 18, 9]
```

## Create Line Plot
This plot shows daily expenses across 7 days. We add a title, axis labels, grid, and save the figure.


```python
plt.figure(figsize=(8, 4))

plt.plot(days, expenses, marker="o")
plt.title("Daily Expenses (7 Days)")
plt.xlabel("Day")
plt.ylabel("Expenses ($)")
plt.grid(True)

plt.savefig("figure_line.png", dpi=150)
plt.show()
```

## Bar Chart — Expenses by Category
Now we will summarize expenses by category and create a bar chart.


```python
categories = ["Food", "Transport", "Internet", "Other"]
values = [35, 20, 15, 10]
```

## Create Bar Chart
This bar chart compares total expenses for each category. We add title, labels, grid, and save the figure.


```python
plt.figure(figsize=(7, 4))

plt.bar(categories, values)
plt.title("Expenses by Category")
plt.xlabel("Category")
plt.ylabel("Total Expenses ($)")
plt.grid(axis="y")

plt.savefig("figure_bar.png", dpi=150)
plt.show()
```
