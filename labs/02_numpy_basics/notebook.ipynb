## Broadcasting بين (10,) و (10,1)

```python
import numpy as np
np.random.seed(42)

a = np.arange(10)          # shape (10,)
b = np.arange(10).reshape(10, 1)   # shape (10,1)

c = a + b                  # broadcasting -> (10,10)

print("a.shape =", a.shape)
print("b.shape =", b.shape)
print("c.shape =", c.shape)
print(c)
```

## np.where

```python
np.random.seed(42)  # حتى تطلع نفس النتائج كل مرة (اختياري)

grades = np.random.randint(0, 101, size=30)  # 30 طالب
status = np.where(grades >= 50, "Pass", "Fail")

print("Grades:", grades)
print("Status:", status)
```

## clip

```python
bad_grades = np.array([20, 55, -10, 130, 99, 101, 0])
fixed_grades = np.clip(bad_grades, 0, 100)

print("Bad:", bad_grades)
print("Fixed:", fixed_grades)
```

## Loop vs Vectorization

```python
import time

N = 1_000_000
x = np.random.randint(0, 101, size=N)

# 1) Loop
t0 = time.time()
loop_out = np.empty_like(x)
for i, v in enumerate(x):
    loop_out[i] = 1 if v >= 50 else 0
t1 = time.time()

# 2) Vectorization
t2 = time.time()
vec_out = (x >= 50).astype(int)
t3 = time.time()

print("Loop time:", round(t1 - t0, 4), "sec")
print("Vector time:", round(t3 - t2, 4), "sec")
print("Same result?", np.array_equal(loop_out, vec_out))
```

## Pass/Fail

```python
grades = np.random.randint(0,101,size=30)
status = np.where(grades >= 50, "pass" , "fail")

passed = np.sum(grades >= 50)
failed = np.sum(grades < 50)
total = grades.size

pass_rate = (passed / total) * 100
fail_rate = (failed / total) * 100

print("=== Small Report ===")
print("Total students:", total)
print("Passed:", passed)
print("Failed:", failed)
print("Pass rate:", round(pass_rate, 2), "%")
print("Fail rate:", round(fail_rate, 2), "%")
```

## Final Summary

```python
import pandas as pd

df = pd.DataFrame({
    "Grade": grades,
    "Status": status
})

passed = (df["Status"] == "pass").sum()
failed = (df["Status"] == "fail").sum()
total = len(df)

pass_pct = (passed / total) * 100
fail_pct = (failed / total) * 100

print("=== Final Report ===")
print("Total students:", total)
print("Passed:", passed)
print("Failed:", failed)
print("Pass %:", round(pass_pct, 2))
print("Fail %:", round(fail_pct, 2))
```
