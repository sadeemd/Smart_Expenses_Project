## 1) Import Libraries
In this section, we import the required Python libraries for data handling and machine learning.


```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, accuracy_score

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import StandardScaler
```

## 2) Load Dataset
We load the dataset that contains at least one numeric column called `amount`.
If no CSV is available, we use sample data for practice.


```python
#Option A: CSV file

#df = pd.read_csv("data.csv")
#df.head()

#Option B: Sample data if you do not have CSV
df = pd.DataFrame({
    "amount": [20, 50, 120, 200, 80, 30, 300, 90, 10, 150, 60, 110, 40, 500]
})
df.head()
```

## 3) Create Label Using Threshold
We create a binary label:
- High if amount >= X
- Low if amount < X
This converts a numeric problem into a classification task.


```python
X_threshold = 100  # Change this value later for the second experiment

df["label"] = np.where(df["amount"] >= X_threshold, "High", "Low")

df[["amount", "label"]].head(10)
```

## 4) Prepare Features and Target
- X (features): the input data (amount)
- y (target): the output label (High / Low)


```python
X = df[["amount"]]   # Feature(s)
y = df["label"]      # Target

print("X shape:", X.shape)
print("Label distribution:\n", y.value_counts())
```

## 5) Train/Test Split
We split the dataset into training and testing sets to evaluate model performance.


```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

print("Train size:", X_train.shape)
print("Test size:", X_test.shape)
```

## 6) Logistic Regression Model
Logistic Regression is a simple and fast classification model.
We scale the feature because Logistic Regression performs better with normalized data.


```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model_lr = LogisticRegression()
model_lr.fit(X_train_scaled, y_train)

y_pred_lr = model_lr.predict(X_test_scaled)

cm_lr = confusion_matrix(y_test, y_pred_lr, labels=["Low", "High"])
acc_lr = accuracy_score(y_test, y_pred_lr)

print("Logistic Regression Results")
print("Confusion Matrix:\n", cm_lr)
print("Accuracy:", acc_lr)
```

## 7) Decision Tree Model
Decision Tree is easy to interpret and does not require scaling.
We train the model and evaluate using confusion matrix and accuracy.


```python
model_dt = DecisionTreeClassifier(random_state=42, max_depth=3)
model_dt.fit(X_train, y_train)

y_pred_dt = model_dt.predict(X_test)

cm_dt = confusion_matrix(y_test, y_pred_dt, labels=["Low", "High"])
acc_dt = accuracy_score(y_test, y_pred_dt)

print("Decision Tree Results")
print("Confusion Matrix:\n", cm_dt)
print("Accuracy:", acc_dt)
```

## 8) Second Threshold Experiment
Now we change the threshold value (X) and repeat the classification.
This helps us see how the label definition affects model performance.


```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import confusion_matrix, accuracy_score, classification_report


# -----------------------------
# DATA (Use sample data)
# If you have a CSV, replace this with:
# df = pd.read_csv("data.csv")
# -----------------------------
df = pd.DataFrame({
    "amount": [20, 50, 120, 200, 80, 30, 300, 90, 10, 150, 60, 110, 40, 500]
})


# -----------------------------
# FUNCTION: Run experiment with a threshold
# -----------------------------
def run_experiment(threshold):
    temp = df.copy()

    # Create labels based on threshold
    temp["label"] = np.where(temp["amount"] >= threshold, "High", "Low")

    # Features and target
    X = temp[["amount"]]
    y = temp["label"]

    # Train / Test split
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.25, random_state=42, stratify=y
    )

    # Scale feature (important for Logistic Regression)
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # Train model
    model = LogisticRegression()
    model.fit(X_train_scaled, y_train)

    # Predict
    y_pred = model.predict(X_test_scaled)

    # Metrics
    cm = confusion_matrix(y_test, y_pred, labels=["Low", "High"])
    acc = accuracy_score(y_test, y_pred)
    report = classification_report(y_test, y_pred, zero_division=0)

    return cm, acc, y.value_counts(), report


# -----------------------------
# RUN TWO THRESHOLDS
# -----------------------------
cm1, acc1, dist1, rep1 = run_experiment(100)
cm2, acc2, dist2, rep2 = run_experiment(150)

print("Threshold = 100")
print("Label distribution:\n", dist1)
print("Confusion Matrix:\n", cm1)
print("Accuracy:", acc1)
print(rep1)

print("\n" + "-"*40 + "\n")

print("Threshold = 150")
print("Label distribution:\n", dist2)
print("Confusion Matrix:\n", cm2)
print("Accuracy:", acc2)
print(rep2)
```
