Pandas Filtering + GroupBy + Aggregation

```python
import pandas as pd
```

#Load Data

```python
df = pd.read_csv("expenses.csv")
df
```

#Filtering (Condition)

```python
filtered = df[df["amount"] >= 15]
filtered
```

Sorting (sort_values)

```python
sorted_df = filtered.sort_values(by="amount", ascending=False)
sorted_df
```

#GroupBy + Aggregation (sum/mean/count)

```python
report = df.groupby("category").agg(
    total_amount=("amount", "sum"),
    avg_amount=("amount", "mean"),
    count=("amount", "count")
).reset_index()

report
```

#Final Sorted Report

```python
final_report = report.sort_values(by="total_amount", ascending=False)
final_report
```

#Compare 3 Categories

```python
selected_categories = ["Food", "Transport", "Bills"]

compare_df = final_report[final_report["category"].isin(selected_categories)]
compare_df
```

#Save Report to CSV

```python
final_report.to_csv("expenses_report.csv", index=False)
print("Saved: expenses_report.csv")
```
