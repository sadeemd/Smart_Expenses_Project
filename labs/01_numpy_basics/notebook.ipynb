```python
import numpy as np
np.__version__
```

```python
a1 = np.array([10, 20, 30, 40, 50])
a2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("a1 =", a1)
print("a2 =\n", a2)

print("a1 shape:", a1.shape)
print("a2 shape:", a2.shape)
```

```python
b_int = np.array([1, 2, 3])
b_float = np.array([1, 2, 3], dtype=float)
b_i16 = np.array([1, 2, 3], dtype=np.int16)

print("b_int dtype:", b_int.dtype)
print("b_float dtype:", b_float.dtype)
print("b_i16 dtype:", b_i16.dtype)
```

```python
print("a1[0] =", a1[0])      # أول عنصر
print("a1[-1] =", a1[-1])    # آخر عنصر

print("a2[0,0] =", a2[0,0])  # صف0 عمود0
print("a2[1,2] =", a2[1,2])  # صف1 عمود2
```

```python
print("a1[1:4] =", a1[1:4])     # من index 1 إلى قبل 4
print("a1[:3]  =", a1[:3])      # أول 3 عناصر
print("a1[::2] =", a1[::2])     # كل عنصرين

print("a2 first row =", a2[0, :])   # كل أعمدة الصف الأول
print("a2 first col =", a2[:, 0])   # كل صفوف العمود الأول
```

```python
idx = [0, 2, 4]
print("a1 fancy =", a1[idx])
```

```python
print("sum a1 =", a1.sum())
print("mean a1 =", a1.mean())

print("sum a2 =", a2.sum())
print("mean a2 =", a2.mean())
```

```python
print("sum by rows =", a2.sum(axis=1))  # مجموع كل صف
print("sum by cols =", a2.sum(axis=0))  # مجموع كل عمود
```

## Challenge: stats(arr) function

```python
def stats(arr):
    arr = np.array(arr)
    avg = float(arr.mean())
    mx = int(arr.max())
    mn = int(arr.min())
    return avg,mx,mn
print(stats(A))
print(stats(M))
```
