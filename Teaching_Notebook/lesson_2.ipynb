## Lesson 2: Practical Data Analysis

Lesson Objective

In this practical lesson, students will learn how to analyze data using basic analytical techniques.
The focus is on understanding key performance indicators (KPIs), creating summary tables, applying filters, and interpreting results instead of just reading numbers.

## What is Data Analysis?

Data analysis is the process of inspecting, cleaning, transforming, and modeling data to discover useful information and support decision-making.
Rather than looking at raw data, analysis helps us answer questions such as:

What is happening?

Why is it happening?

What should we do next?

## Key Performance Indicators (KPIs)

KPIs are measurable values that show how effectively a process or system is performing.
In this lesson, we will use simple KPIs such as:

Average value

Maximum value

Minimum value

Total count

These indicators help summarize the data in a clear and meaningful way.

## Grouping Data (Aggregation Tables)

Grouping data allows us to analyze information by category.
For example:

Average score per subject

Total values per group

Using aggregation helps us compare categories and identify patterns or differences.

## Filtering Data

Filtering means selecting specific rows that meet certain conditions.
Examples include:

Scores greater than a specific value

Records related to a specific category

Filtering helps focus the analysis on relevant data only.

## Interpreting Results

Numbers alone are not enough.
After every analysis, we must ask:

What does this result mean?

Is the performance good or bad?

What decision can be made based on this result?

Interpretation is the most important step in data analysis.

## Student Activity

Students are required to:

Complete the provided KPI tables

Analyze grouped data

Apply filters

Identify the Top 5 values based on sorting

No solutions are provided; students must calculate and interpret results on their own.

## Conclusion

By the end of this lesson, students will be able to:

Convert raw data into useful information

Use KPIs and aggregation techniques effectively

Apply basic data analysis skills in real-world scenarios
